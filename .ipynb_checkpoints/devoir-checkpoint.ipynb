```python
import seaborn as sns
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 

# ================================================
# 1. CHARGEMENT ET NETTOYAGE DES DONNÉES
# ================================================

print("|--- Chargement et Nettoyage de la base Auto MPG ---|")
df_raw = sns.load_dataset('mpg')

# print(df_raw)
# Retrait de la colonne 'name' 
df_clean = df_raw.drop(columns=['name'])

# Suppression des valeurs manquantes 
df_clean = df_clean.dropna().reset_index(drop=True)

print(f"Dimensions initiales: {df_raw.shape}")
print(f"Dimensions apres nettoyage: {df_clean.shape}")
print(df_clean)

```

```python
# ==================================
# 2. STRATÉGIE DE DISCRÉTISATION 
# ==================================


df_clean['mpg_cut'] = pd.cut(
    df_clean['mpg'],
    bins=3,
    labels=['faible', 'moyenne', 'elevee']
)

df_clean['mpg_qcut'], seuils_quantiles = pd.qcut(
    df_clean['mpg'],
    q=3,
    labels=['faible', 'moyenne', 'elevee'],
    retbins=True
)

print('Comparaison de la repartition des classes')

print("1. Repartition avec tranches fixes 'cut' ")
print(df_clean['mpg_cut'].value_counts())

print("\n2. Répartition avec quantiles 'qcut' ")
print(df_clean['mpg_qcut'].value_counts())

print(f"Seuils retenus pour les quantiles: {np.round(seuils_quantiles, 2)}\n")
```

```python
# ===================================================
# BLOC 2 : LE PIPELINE DE PRETRAITEMENT JUSTIFIE
# ===================================================
from sklearn.model_selection import train_test_split 

print("|--- Preparation des matrices X et y pour le SVM ---|")


y = df_clean['mpg_qcut'].values

# On retire les colonnes non explicatives (mpg continu, mpg_cut et mpg_qcut)
X_brut = df_clean.drop(columns=['mpg', 'mpg_cut', 'mpg_qcut'])

X_encode = pd.get_dummies(X_brut, columns=['origin'], drop_first=True)

# print(X_encode)

X_train_raw, X_test_raw, y_train, y_test = train_test_split(
    X_encode, y, test_size=0.2, random_state=42, stratify=y
)
moyenne_train = X_train_raw.mean(axis=0)
std_train = X_train_raw.std(axis=0)

X_train = ((X_train_raw - moyenne_train) / std_train).values
X_test = ((X_test_raw - moyenne_train) / std_train).values 

print("Pretraitement termine avec succes !")
print(f"Taille de X_train: {X_train.shape}")
print(f"Taille de X_test: {X_test.shape}")

print(f"Les 3 classes cibles : {np.unique(y_train)}")
```

```python
# ============================
# BLOC 3 (SIMPLIFIÉ) : 'w⋅x+b' 
# ============================
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print('|--- Entrainement du SVC Lineaire ---|')

def entrainer_svm_binaire(X, y_bin, C=1.0, lr=0.005, epochs=1500):
    n_echantillons, n_variables = X.shape 
    w = np.zeros(n_variables)
    b = 0.0 
    
    for _ in range(epochs):
        for i in range(n_echantillons):
            # Formule de décision : y_i * (w . x_i + b)
            marge = y_bin[i] * (np.dot(X[i], w) + b)
            
            # Si le point est mal classé ou dans la marge (marge < 1)
            if marge < 1.0: 
                w -= lr * ((w / epochs) - C * y_bin[i] * X[i])
                b += lr * C * y_bin[i]
            else:
             # Si le point est bien classé hors de la marge
                w -= lr * (w /epochs)

    return w, b

# 2. Fonction qui gère les 3 classes ('faible', 'moyenne', 'elevee')
def entrainer_svm_ovr(X, y, C=1.0, lr=0.005, epochs=1500):
    classes = np.unique(y)
    modeles = {} 

    for c in classes:
        y_binaire = np.where(y == c, 1.0, -1.0)
        w, b = entrainer_svm_binaire(X, y_binaire, C=C, lr=lr, epochs=epochs) 
        modeles[c] = (w, b)

    return modeles

# Fonction pour faire la prediction
def predire_svm(modeles, X):
    classes = list(modeles.keys())
    scores = []

    # Pour chaque classe, on calcule le score : w . x + b
    for c in classes: 
        w, b = modeles[c]
        score_c = np.dot(X, w) + b
        scores.append(score_c)

    scores = np.array(scores).T
    meilleurs_indices = np.argmax(scores, axis=1)

    return np.array([classes[idx] for idx in meilleurs_indices])

modeles_lineaires = entrainer_svm_ovr(X_train, y_train, C=1.0, lr=0.005, epochs=1500)

y_pred_linear = predire_svm(modeles_lineaires, X_test)

acc = accuracy_score(y_test, y_pred_linear)
print(f"Precision du SVC Lineaire : {acc * 100:.2f} %\n")
print("Rapport détaillé de classification :")
print(classification_report(y_test, y_pred_linear))
```

```python
# =============================
# BLOC 4 : SVC AVEC NOYAU RBF 
# =============================
 
print("|--- Entrainement du SVC avec Noyeau RBF From Scratch ---|")

def calculer_noyau_rbf(X1, X2, gamma=0.1):
    """
    Calcule la similarite entre chaque voiture de X1 et chaque voiture de X2.
    """
    # Calcul rapide et vectorisé des distances au carré entre tous les points
    X1_carre = np.sum(X1**2, axis=1)[:, np.newaxis]
    X2_carre = np.sum(X2**2, axis=1)
    distances_carre = X1_carre + X2_carre - 2 * np.dot(X1, X2.T)

    return np.exp(-gamma * distances_carre)

gamma_choisi = 0.1 

K_train = calculer_noyau_rbf(X_train, X_train, gamma=gamma_choisi)
K_test = calculer_noyau_rbf(X_test, X_train, gamma=gamma_choisi)

# 3. Entraînement du SVM sur la matrice de ressemblance RBF
modeles_rbf = entrainer_svm_ovr(K_train, y_train, C=1.5, lr=0.005, epochs=1500)

y_pred_rbf = predire_svm(modeles_rbf, K_test)

acc_rbf = accuracy_score(y_test, y_pred_rbf)

print(f"Precision du SVC RBF 'FROM SCRATCH': {acc_rbf * 100:.2f} %")
print(f"Amelioration par rapport au modele Lineaire: +{(acc_rbf - acc) * 100:.2f} %\n")
print(f"Rapport de classification RBF :")
print(classification_report(y_test, y_pred_rbf))
```

```python
# ========================================================
# BLOC 5 : COMPARAISON SCIKIT-LEARN ET VISUALISATIONS
# ========================================================
from sklearn.svm import SVC
import seaborn as sns 

print("|--- Comparaison avec Scikit-Learn & Trace Les Graphiques ---|")

sklearn_linear = SVC(kernel="linear", C=1.0, random_state=42)
sklearn_linear.fit(X_train, y_train)
acc_sk_linear = accuracy_score(y_test, sklearn_linear.predict(X_test))

sklearn_rbf = SVC(kernel='rbf', C=1.5, gamma=0.1, random_state=42)
sklearn_rbf.fit(X_train, y_train)
acc_sk_rbf = accuracy_score(y_test, sklearn_rbf.predict(X_test))


tableau_comparatif = pd.DataFrame({
    'Modele': ['SVC Lineare', 'SVC a Noyau RBF'],
    'Notre Code': [f"{acc * 100:.2f} %", f"{acc_rbf * 100:.2f} %"],
    "Scikit-Learn 'Industrie'": [f'{acc_sk_linear * 100:.2f} %', f"{acc_sk_rbf * 100:.2f} %"]
})

print("\n--- Tableau Comparatif Final 'Classification' ---")
print(tableau_comparatif.to_string(index=False))
print("==================================================\n")

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Graphique 1 : Comparaison des Précisions

labels = ['Lineaire', 'Noyau RBF']
scratch_scores = [acc * 100, acc_rbf * 100]
sk_scores = [acc_sk_linear * 100, acc_sk_rbf * 100]

x = np.arange(len(labels))
largeur = 0.35 

axes[0].bar(x - largeur/2, scratch_scores, largeur, label='From Scratch (Notre code)', color='#1f77b4', edgecolor='black')
axes[0].bar(x + largeur/2, sk_scores, largeur, label='Scikit-Learn', color='#2ca02c', edgecolor='black')
axes[0].set_ylabel('Précision (Accuracy en %)')
axes[0].set_title('Comparaison : From Scratch vs Scikit-Learn', fontsize=11)
axes[0].set_xticks(x)
axes[0].set_xticklabels(labels)
axes[0].set_ylim(60, 100)
axes[0].legend()
axes[0].grid(axis='y', alpha=0.3)

# Graphique 2 : Matrice de Confusion du modèle RBF
mat_conf = confusion_matrix(y_test, y_pred_rbf, labels=['faible', 'moyenne', 'elevee'])
sns.heatmap(mat_conf, annot=True, fmt='d', cmap='Blues', ax=axes[1],
            xticklabels=['Faible', 'Moyenne', 'Élevée'],
            yticklabels=['Faible', 'Moyenne', 'Élevée'])
axes[1].set_title('Matrice de Confusion - SVC RBF (From Scratch)', fontsize=11)
axes[1].set_xlabel('Prédiction')
axes[1].set_ylabel('Vérité Réelle')

plt.tight_layout()
plt.show()
```

```python
# ========================
# BLOC 6 : SVR LINEAIRE
# ========================
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score 

print('Entrainement du SVR Lineaire From Scratch')
y_continu = df_clean['mpg'].values 

y_train_reg = y_continu[X_train_raw.index]
y_test_reg = y_continu[X_test_raw.index]

# 2. Fonc. d'entrainement du SVR lineaire
def entrainer_svr_lineaire(X, y, epsilon=1.5, C=3.0, lr=0.01, epochs=2500):
    """
    Entraîne un SVR lineaire avec descente de sous-gradient sur la perte epsilon-insensible.
    """
    n_samples, n_features = X.shape
    w = np.zeros(n_features)
    b = float(np.mean(y)) 

    for epoch in range(epochs):
         
        lr_actuel = lr / (1.0 + 0.001 * epoch)
        for i in range(n_samples):
            # Prédiction actuelle : w . x_i + b
            prediction = np.dot(X[i], w) + b
            erreur = prediction - y[i]
            
            # Cas 1 : Point DANS le tube (|erreur| <= epsilon) -> perte = 0
            if abs(erreur) <= epsilon:
                w -= lr_actuel * (0.01 * w)
            
            # Cas 2 : Point AU-DESSUS du tube (erreur > epsilon)
            elif erreur > epsilon:
                w -= lr_actuel * (0.01 * w + C * X[i])
                b -= lr_actuel * C
            
            # Cas 3 : Point EN-DESSOUS du tube (erreur < -epsilon)
            else:
                 w -= lr_actuel * (0.01 * w - C * X[i])
                 b += lr_actuel * C
        
        return w, b

# Fonction de prediction
def predire_svr(w, b, X):
    return np.dot(X, w) + b


# Fonction pour identifier les Vecteurs de Support
def trouver_vecteurs_support(w, b, X, y, epsilon=1.5):
    predictions = predire_svr(w, b, X)
    erreurs = np.abs(y - predictions)

    indices_sv = np.where(erreurs >= (epsilon - 1e-4))[0]    
    return indices_sv

# =========================================
# ENTRAINEMENT DU MODELE AVEC EPSILON 
# =========================================
epsilon_test = 1.5 
w_svr, b_svr = entrainer_svr_lineaire(X_train, y_train_reg, epsilon=epsilon_test, C=3.0, lr=0.01, epochs=2500)

# Predictions sur le jeu de test
y_pred_svr = predire_svr(w_svr, b_svr, X_test)

mse_svr = mean_squared_error(y_test_reg, y_pred_svr)
rmse_svr = np.sqrt(mse_svr)
mae_svr = mean_absolute_error(y_test_reg, y_pred_svr)
r2_svr = r2_score(y_test_reg, y_pred_svr)

sv_indices = trouver_vecteurs_support(w_svr, b_svr, X_train, y_train_reg, epsilon=epsilon_test)

print(f"Modele SVR entraine avec epsilon = {epsilon_test} mpg")
print(f"Nbre de vecteurs de support : {len(sv_indices)} sur {len(X_train)} '{len(sv_indices)/len(X_train)*100:.1f} %'")
print(f"Erreur moy. absolue 'MAE': {mae_svr:.2f} mpg")
print(f"Racine de l'erreur quadratique 'RMSE': {rmse_svr:.2f} mpg")
print(f"Coefficient de det 'R2': {r2_svr:.4f}")
```

```python
# =====================================================================
# BLOC 7 : ÉTUDE DE L'INFLUENCE DE EPSILON & COMPARAISON SCIKIT-LEARN
# =====================================================================
from sklearn.svm import SVR 

print("Etape 7 : Etude de l'influence de Epsilon et Comparaison")
vals_epsilon = [0.2, 0.5, 1.0, 1.5, 2.5, 4.0, 6.0]
resultats_etude = [] 

for eps in vals_epsilon:
    w_tmp, b_tmp = entrainer_svr_lineaire(X_train, y_train_reg,  epsilon=eps)
    y_pred_tmp = predire_svr(w_tmp, b_tmp, X_test)

    sv_tmp = trouver_vecteurs_support(w_tmp, b_tmp, X_train, y_train_reg, epsilon=eps)
    pct_sv = (len(sv_tmp) / len(X_train)) * 100

    mae_tmp = mean_absolute_error(y_test_reg, y_pred_tmp)
    r2_tmp = r2_score(y_test_reg, y_pred_tmp)

    resultats_etude.append({
        'Epsilon': eps, 
        'Nb Vecteurs Support': len(sv_tmp),
        '% Vecteurs Support': f"{pct_sv:.1f} %",
        'MAE Test (mpg)': round(mae_tmp, 2),
        'R2 Test': round(r2_tmp, 4)
    })

df_etude = pd.DataFrame(resultats_etude)
print("\n=== Tableau : Impact de epsilon sur les vecteurs de support et le bruit ---")
print(df_etude.to_string(index=False))
print("==============================================================================")

sklearn_svr = SVR(kernel='linear', C=3.0, epsilon=1.5)
sklearn_svr.fit(X_train, y_train_reg)
y_pred_sk_svr = sklearn_svr.predict(X_test)

mae_sk_svr = mean_absolute_error(y_test_reg, y_pred_sk_svr)
r2_sk_svr = r2_score(y_test_reg, y_pred_sk_svr)
nb_sv_sk = len(sklearn_svr.support_)

print("----- Comparaison from Scratch - SCIKIT-LEARN 'EPSILON = 1.5' -----")
print(f"Notre Code 'From Scratch': R2 = {r2_svr:.4f} | MAE = {mae_svr:.2f} mpg | SV = {len(sv_indices)} '{len(sv_indices)/len(X_train)*100:.1f}%'")
print(f"Scikit-Learn officiel : R2 = {r2_sk_svr:.4f} | MAE = {mae_sk_svr:.2f} mpg | SV = {nb_sv_sk} '{nb_sv_sk/len(X_train)*100:.1f}%'")
print("=================================================================\n")

# ===========================
# 3. TRACE DES DEUX DESSINS 
# ===========================
fig, axes = plt.subplots(1, 2, figsize=(15, 6))


# Graphique A : Evolution du nombre de vecteurs de support selon Epsilon

eps_vals = [r['Epsilon'] for r in resultats_etude]
nb_sv_vals = [r['Nb Vecteurs Support'] for r in resultats_etude]

axes[0].plot(eps_vals, nb_sv_vals, marker='o', linewidth=2.5, color='#d62728')
axes[0].set_title("Influence de Epsilon sur le Nombre de Vecteurs de Support", fontsize=11)
axes[0].set_xlabel("Largeur du tube Epsilon (mpg)", fontsize=10)
axes[0].set_ylabel("Nombre de Vecteurs de Support (sur 313)", fontsize=10)
axes[0].grid(True, linestyle='--', alpha=0.5)

axes[0].annotate("Petit epsilon : presque tous les points\nsont des vecteurs de support (sensible au bruit)",
                 xy=(eps_vals[0], nb_sv_vals[0]), xytext=(0.8, nb_sv_vals[0] - 20),
                 arrowprops=dict(arrowstyle="->", color="black"), fontsize=9)

axes[0].annotate("Grand epsilon : tube très large,\nmodèle très parcimonieux (ignore le bruit)",
                 xy=(eps_vals[-1], nb_sv_vals[-1]), xytext=(3.0, nb_sv_vals[-1] + 30),
                 arrowprops=dict(arrowstyle="->", color="black"), fontsize=9)

# Graphique B : Visualisation du Tube Epsilon-insensible
erreurs_test = np.abs(y_test_reg - y_pred_svr)
dans_tube = erreurs_test <= epsilon_test

axes[1].scatter(y_test_reg[dans_tube], y_pred_svr[dans_tube], color='#2ca02c', alpha=0.8,
                label=f'Dans le tube (Erreur <= {epsilon_test} : Bruit ignoré)')
axes[1].scatter(y_test_reg[~dans_tube], y_pred_svr[~dans_tube], color='#d62728', alpha=0.8,
                label=f'Hors du tube (Erreur > {epsilon_test} : Vecteurs Support)')


lim_x = np.linspace(min(y_test_reg) - 2, max(y_test_reg) + 2, 100)
axes[1].plot(lim_x, lim_x, 'k-', linewidth=1.5, label='Prédiction idéale (y = x)')
axes[1].plot(lim_x, lim_x + epsilon_test, 'k--', linewidth=1, label=f'Frontières du tube (+/- {epsilon_test} mpg)')
axes[1].plot(lim_x, lim_x - epsilon_test, 'k--', linewidth=1)
axes[1].fill_between(lim_x, lim_x - epsilon_test, lim_x + epsilon_test, color='gray', alpha=0.15)

axes[1].set_title(f"Visualisation du Tube Epsilon-insensible (SVR Lineaire)", fontsize=11)
axes[1].set_xlabel("Vraie consommation MPG (Réalité)", fontsize=10)
axes[1].set_ylabel("Consommation MPG prédite", fontsize=10)
axes[1].legend(fontsize=8, loc='upper left')
axes[1].grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
